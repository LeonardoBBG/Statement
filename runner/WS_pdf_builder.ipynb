```python
from pathlib import Path
import subprocess
import json

# We are running from: project_root/runner/
RUNNER_DIR = Path.cwd().resolve()
ROOT = RUNNER_DIR.parent

CODE_DIR = ROOT / "code"
INPUT_DIR = ROOT / "input"

PY_SCRIPT = CODE_DIR / "fate_columns_from_pdf.py"

assert CODE_DIR.exists(), f"Missing: {CODE_DIR}"
assert INPUT_DIR.exists(), f"Missing: {INPUT_DIR}"
assert PY_SCRIPT.exists(), f"Missing: {PY_SCRIPT}"

print("ROOT:", ROOT)
print("CODE_DIR:", CODE_DIR)
print("INPUT_DIR:", INPUT_DIR)
print("PY_SCRIPT:", PY_SCRIPT)


# Change this only if the PDF name changes
FATE_PDF = INPUT_DIR / "20260111_144606__ws_claimant_analysis_narrative_qwen2.532b.pdf"

assert FATE_PDF.exists(), f"Missing FATE PDF: {FATE_PDF}"

OUT_DIR = ROOT / "output"  # write outputs next to inputs (intentional)

print("FATE_PDF:", FATE_PDF)
print("OUT_DIR:", OUT_DIR)


cmd = [
    "python",
    str(PY_SCRIPT),
    "--pdf", str(FATE_PDF),
    "--out_dir", str(OUT_DIR),
]

print("Running command:")
print(" ".join(cmd))

result = subprocess.run(
    cmd,
    capture_output=True,
    text=True,
)

print("STDOUT:")
print(result.stdout)

print("STDERR:")
print(result.stderr)

assert result.returncode == 0, "Extractor failed"


fate_json = OUT_DIR / "fate_columns.json"
fate_csv = OUT_DIR / "fate_columns.csv"

assert fate_json.exists(), "Missing fate_columns.json"
assert fate_csv.exists(), "Missing fate_columns.csv"

with open(fate_json, "r", encoding="utf-8") as f:
    data = json.load(f)

print("Extracted metadata:")
print(data["meta"])

print("\nClusters:")
for c in data["clusters"]:
    print("-", c)

print("\nKEGs:")
for k in data["kegs"]:
    print("-", k["id"], ":", k["title"])
```
